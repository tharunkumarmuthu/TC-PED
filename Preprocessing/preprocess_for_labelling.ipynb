```python
#To merge multiple scrapeds tweet data CSV files into one
import pandas as pd

# List of CSV files to merge
files = [
    'tn_political_data_full.csv',
    'tn_political_data_full_2.csv',
    'tn_political_data_full_latest.csv',
    'tn_political_data_full_latest_2.csv'
]

# Read all CSVs into a list of DataFrames
dfs = [pd.read_csv(f) for f in files]

# Concatenate all DataFrames
combined_df = pd.concat(dfs, ignore_index=True)

# Drop duplicates based on 'tweet_id'
combined_df.drop_duplicates(subset='tweet_id', inplace=True)

# Save the merged DataFrame to a new CSV
combined_df.to_csv('merged_tweets.csv', index=False)

print("Merged CSV created: merged_tweets.csv")
print(f"Total rows in merged CSV: {len(combined_df)}")
```

```python
# To Create  a csv for Label Studio to label the tweets using the merged tweets
df = pd.read_csv("merged_tweets.csv")
```

```python
import re

def make_clickable(text):
    # Regex to find URLs starting with http or https
    url_pattern = r'(https?://[^\s]+)'
    # Replace the URL with a clickable HTML link
    # target="_blank" ensures the link opens in a new tab
    return re.sub(url_pattern, r'<a href="\1">\1</a>', str(text))

# Apply the function to your dataframe
df['full_text'] = df['full_text'].apply(make_clickable)
```

```python
def addurl(text):
    text = str(text)

    return f'<a href="https://x.com/twitter/status/{text}" target="_blank">Tweet Link</a>'

# Apply the function to your dataframe
df['link'] = df['tweet_id'].apply(addurl)
```
